```python
import proveit
# Prepare this notebook for defining the theorems of a theory:
%theorems_notebook # Keep this at the top following 'import proveit'.

from proveit import k, p
from proveit.logic import Forall
from proveit.numbers import two, Exp, NaturalPos, Prime, Rational, Real, Complex, Add, Mult
from proveit.abstract_algebra import FiniteField, IsField
```

```python
%begin theorems
```

```python
rational_field = IsField(Rational, Add._operator_, Mult._operator_)
```

```python
real_field = IsField(Real, Add._operator_, Mult._operator_)
```

```python
complex_field = IsField(Complex, Add._operator_, Mult._operator_)
```

```python
# a place-holder until we can be more accurate
# two_elem_finite_field = InClass(FiniteField(two), Fields(Add._operator_, Mult._operator_))
two_elem_finite_field = IsField(FiniteField(two), Add._operator_, Mult._operator_)
```

```python
# A more general theorem: 
# Forall primes p and forall natpos k, FiniteField(p^{k}) will be a field over the
# appropriate modular addition and multiplication
finite_fields_order_p_k = Forall(
        (p, k), IsField(FiniteField(Exp(p, k)),
                        Add._operator_, Mult._operator_),
        domains = [Prime, NaturalPos]
)
```
