```python
from proveit import a, b, c, q, u, v, A, B, C, G, H, P, X, Function
from proveit.logic import And, Equals, Exists, Forall, Implies, InSet
from proveit.logic.sets.partitions import Partitions
from proveit.logic.sets import Bijections, Difference, IsBijection, Set, Surjections
from proveit.graphs import P_G, P_H, EndVertices, Graphs, IsGraph, IsPath, PathsOf, QuotientGraph, Vertices
```

```python
path_preservation = (
Forall((G, H),
Forall((A, B, C),
Forall(q,
Forall((a, b),
       Implies(And(Forall(v, Equals(Function(q, v), a), domain=A),
                   Forall(v, Equals(Function(q, v), b), domain=B),
                   IsBijection(q, C, Difference(Vertices(H), Set(a, b)))),
       Implies(Exists(P_H, InSet(P_H, PathsOf(H, ends=(a,b)))),
               Exists((u, v),
                      Exists(P_G, InSet(P_G, PathsOf(G, ends=(u,v)))),
                      domains=[A, B])
              )).with_wrap_before_operator(),
domain=Vertices(H)).with_wrapping(),
conditions=[Equals(H, QuotientGraph(G, q))]).with_wrapping(),
conditions=[InSet(Set(A, B, C), Partitions(Vertices(G)))]).with_wrapping(),
conditions=[IsGraph(G)]).with_wrapping()
)
```
