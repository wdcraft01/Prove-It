```python
import proveit
# Prepare this notebook for defining the theorems of a theory:
%theorems_notebook # Keep this at the top following 'import proveit'.

# from ... import ...
from proveit import E, G, V
from proveit.logic import Equals, Forall
from proveit.logic.sets import CartExp, SubsetEq
from proveit.numbers import two
from proveit.graphs import Digraph, Edges, IsDigraph, Vertices
```

```python
%begin theorems
```

### Digraphs & Digraph(V, E)

```python
is_digraph_unfolding = (
    Forall(G, SubsetEq(Edges(G), CartExp(Vertices(G), two)),
          conditions = [IsDigraph(G)])
)
```

```python
is_digraph_folding = (
    Forall(G, IsDigraph(G),
          conditions = [SubsetEq(Edges(G), CartExp(Vertices(G), two))])
)
```

```python
digraph_v_e_is_digraph_def = (
    Forall((V,E), Equals(IsDigraph(Digraph(V,E)),
                        SubsetEq(E, CartExp(V, two)))
    )
)
```

```python
digraph_v_e_is_digraph_unfolding = (
    Forall((V,E), SubsetEq(E, CartExp(V, two)),
    conditions=[IsDigraph(Digraph(V,E))]
    )
)
```

```python
digraph_v_e_is_digraph_folding = (
    Forall((V,E), IsDigraph(Digraph(V,E)),
    conditions=[SubsetEq(E, CartExp(V, two))]
    )
)
```

### Adjacency

```python
from proveit import u, v
from proveit.logic import InSet
from proveit.graphs import AdjacentVertices
adjacent_vertices_digraph_unfolding = (
    Forall(G,
           Forall((u, v),
                  InSet((u, v), Edges(G)),
                  conditions=[AdjacentVertices(u, v, G)], domain = Vertices(G)),
           conditions = [IsDigraph(G)]
    )
)
```

```python
adjacent_vertices_digraph_folding = (
    Forall(G,
           Forall((u, v),
                  AdjacentVertices(u, v, G),
                  conditions=[InSet((u, v), Edges(G))], domain = Vertices(G)),
           conditions = [IsDigraph(G)]
    )
)
```
