```python
import proveit
from proveit import defaults, display_provers # useful imports
from proveit import E, G, V
from proveit.logic.sets import CartExp, SubsetEq
from proveit.graphs import Edges, Digraph, IsDigraph, Vertices
from proveit.numbers import two
%begin demonstrations
```

### Digraphs and Related Constructions & Methods $G = \textrm{Digraph}(V, E)$

#### Miscellaneous Testing
Testing various classes and methods related to the class of digraphs and the representation of a digraph $G = (V, E)$ consisting of a set $V$ of vertices and a set $E$ of edges. Some of this material could eventually be integrated into the `_demonstrations_` page and/or deleted as development continues.

```python
example_digraph = Digraph(V, E)
```

```python
IsDigraph(example_digraph)
```

```python
IsDigraph(example_digraph).definition()
```

```python
IsDigraph(example_digraph).as_defined()
```

```python
IsDigraph(example_digraph).unfold(assumptions=[IsDigraph(example_digraph)])
```

```python
IsDigraph(example_digraph).conclude(
    assumptions=[SubsetEq(Edges(example_digraph), CartExp(Vertices(example_digraph), two))])
```

```python
IsDigraph(G).conclude(assumptions=[SubsetEq(Edges(G), CartExp(Vertices(G), two))])
```

```python
IsDigraph(example_digraph).derive_is_graph(assumptions=[IsDigraph(example_digraph)])
```

```python
IsDigraph(example_digraph).derive_size_in_natural(assumptions=[IsDigraph(example_digraph)])
```
