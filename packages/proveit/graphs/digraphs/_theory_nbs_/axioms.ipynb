```python
import proveit
# Prepare this notebook for defining the axioms of a theory:
%axioms_notebook # Keep this at the top following 'import proveit'.

# from ... import ...
from proveit import u, v, G, X
from proveit.logic import Equals, Forall, InSet
from proveit.logic.sets import CartExp, SubsetEq
from proveit.graphs import AdjacentVertices, Edges, IsDigraph, IsGraph, Vertices
from proveit.numbers import two 
```

The expression $\text{Digraph}(V, E)$ is essentially a representation of the ordered pair $(V, E)$, considered as a _possible_ digraph. For $\text{Digraph}(V, E)$ to be a _legitimate_ digraph, we must have that the set $E$ of edges is a subset of the Cartesian product V x V, i.e., $E$ is a subset of the set of all ordered tuples of $(u, v)$ for $u, v \in V$. This observation is captured in the `is_digraph_def` axiom below:

```python
is_digraph_def = (
    Forall(G, Equals(IsDigraph(G), SubsetEq(Edges(G), CartExp(Vertices(G), two))))
)
```

#### A Digraph is a Graph
By definition, a digraph is a graph. This then gives us access to properties and methods of graphs more generally, such as definitions and facts about vertices and edges.

```python
digraph_is_graph = Forall(G, IsGraph(G), conditions=[IsDigraph(G)])
```

#### Adjacency in Digraphs

```python
adjacent_vertices_digraph_def = (
    Forall(G,
           Forall((u, v),
                  Equals(AdjacentVertices(u, v, G), InSet((u, v), Edges(G))),
                  domain = Vertices(G)),
           conditions = [IsDigraph(G)]
    )
)
```
