```python
import proveit
# Prepare this notebook for defining the axioms of a theory:
%axioms_notebook # Keep this at the top following 'import proveit'.

# from ... import ...
from proveit import u, v, G
from proveit.logic import Equals, Forall, InSet
from proveit.logic.sets import KPowerSet, Set, SubsetEq
from proveit.graphs import AdjacentVertices, Edges, IsGraph, IsUndirectedGraph, Vertices
from proveit.numbers import two
```

```python
%begin axioms
```

The expression $\text{UndirectedGraph}(V, E)$ is essentially a representation of the ordered pair $(V, E)$, considered as a _possible_ undirected graph. For $\text{UndirectedGraph}(V, E)$ to be a _legitimate_ undirected graph, we must have that the set $E$ of edges is a subset of the set $[V]^{2}$ of 2-element subsets (not ordered pairs) of vertex set $V$. This observation is captured in the `is_undirected_graph_def` axiom below:

```python
is_undirected_graph_def = (
    Forall(G, Equals(IsUndirectedGraph(G), SubsetEq(Edges(G), KPowerSet(Vertices(G), two))))
)
```

#### An Undirected Graph is a Graph
By definition, an undirected graph is a graph. This then gives us access to properties and methods of graphs more generally, such as definitions and facts about vertices and edges.

```python
undirected_graph_is_graph = Forall(G, IsGraph(G), conditions=[IsUndirectedGraph(G)])
```

#### Adjacency in Undirected Graphs

```python
adjacent_vertices_undirected_graph_def = (
    Forall(G,
           Forall((u, v),
                  Equals(AdjacentVertices(u, v, G), InSet(Set(u, v), Edges(G))),
                  domain = Vertices(G)),
           conditions = [IsUndirectedGraph(G)]
    )
)
```
