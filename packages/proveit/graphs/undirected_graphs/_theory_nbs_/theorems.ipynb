```python
import proveit
# Prepare this notebook for defining the theorems of a theory:
%theorems_notebook # Keep this at the top following 'import proveit'.

# from ... import ...
from proveit import E, G, V
from proveit.logic import Equals, Forall
from proveit.logic.sets import KPowerSet, SubsetEq
from proveit.graphs import Edges, IsUndirectedGraph, UndirectedGraph, Vertices
from proveit.numbers import two
```

### Undirected Graphs & UndirectedGraph(V, E)

```python
is_undirected_graph_unfolding = (
    Forall(G, SubsetEq(Edges(G), KPowerSet(Vertices(G), two)),
          conditions=[IsUndirectedGraph(G)])
)
```

```python
is_undirected_graph_folding = (
    Forall(G, IsUndirectedGraph(G),
          conditions=[SubsetEq(Edges(G), KPowerSet(Vertices(G), two))])
)
```

```python
undirected_graph_v_e_is_undirected_graph_def = (
    Forall((V,E), Equals(IsUndirectedGraph(UndirectedGraph(V,E)),
                        SubsetEq(E, KPowerSet(V, two)))
    )
)
```

```python
undirected_graph_v_e_is_undirected_graph_unfolding = (
    Forall((V,E), SubsetEq(E, KPowerSet(V, two)),
    conditions=[IsUndirectedGraph(UndirectedGraph(V,E))]
    )
)
```

```python
undirected_graph_v_e_is_undirected_graph_folding = (
    Forall((V,E), IsUndirectedGraph(UndirectedGraph(V,E)),
    conditions=[SubsetEq(E, KPowerSet(V, two))]
    )
)
```

### Adjacency

```python
from proveit import u, v
from proveit.logic import InSet
from proveit.logic.sets import Set
from proveit.graphs import AdjacentVertices
adjacent_vertices_undirected_graph_unfolding = (
    Forall(G,
           Forall((u, v),
                  InSet(Set(u, v), Edges(G)),
                  conditions=[AdjacentVertices(u, v, G)], domain = Vertices(G)),
           conditions = [IsUndirectedGraph(G)]
    )
)
```

```python
adjacent_vertices_undirected_graph_folding = (
    Forall(G,
           Forall((u, v),
                  AdjacentVertices(u, v, G),
                  conditions=[InSet(Set(u, v), Edges(G))], domain = Vertices(G)),
           conditions = [IsUndirectedGraph(G)]
    )
)
```
