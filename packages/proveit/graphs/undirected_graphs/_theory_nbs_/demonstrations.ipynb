### Undirected Graphs and Related Constructions & Methods $G = \textrm{UndirectedGraph}(V, E)$

#### Miscellaneous Testing
Testing various classes and methods related to the class of digraphs and the representation of an undirected graph $G = (V, E)$ consisting of a set $V$ of vertices and a set $E$ of edges. Some of this material could eventually be integrated into the `_demonstrations_` page and/or deleted as development continues.

```python
example_undigraph = UndirectedGraph(V, E)
```

```python
IsUndirectedGraph(example_undigraph)
```

```python
IsUndirectedGraph(example_undigraph).definition()
```

```python
IsUndirectedGraph(example_undigraph).as_defined()
```

```python
IsUndirectedGraph(example_undigraph).unfold(assumptions=[IsUndirectedGraph(example_undigraph)])
```

```python
IsUndirectedGraph(example_undigraph).conclude(
    assumptions=[SubsetEq(Edges(example_undigraph), KPowerSet(Vertices(example_undigraph), two))])
```

```python
IsUndirectedGraph(G).conclude(
    assumptions=[SubsetEq(Edges(G), KPowerSet(Vertices(G), two))])
```

```python
IsUndirectedGraph(example_undigraph).derive_is_graph(assumptions=[IsUndirectedGraph(example_undigraph)])
```

```python
IsUndirectedGraph(G).derive_is_graph(assumptions=[IsUndirectedGraph(G)])
```

```python
IsUndirectedGraph(example_undigraph).derive_size_in_natural(assumptions=[IsUndirectedGraph(example_undigraph)])
```
