Theorems (or conjectures) for the theory of <a class="ProveItLink" href="theory.ipynb">proveit.logic.sets.symmetric_difference</a>
========

```python
import proveit
# Prepare this notebook for defining the theorems of a theory:
%theorems_notebook # Keep this at the top following 'import proveit'.

from proveit import i, l, m, n, x, A, B, C, X, ExprRange, IndexedVar
from proveit.core_expr_types import (
        A_1_to_l, A_1_to_m, A_1_to_n, B_1_to_m, B_1_to_n, C_1_to_m, C_1_to_n, D_1_to_n)
from proveit.logic import Equals, Forall, InSet, XOr
from proveit.logic.sets import Difference, EmptySet, Intersect, SymmetricDifference, Union 
from proveit.numbers import one, Natural, NaturalPos
```

```python
%begin theorems
```

```python
membership_unfolding = (
    Forall(n,
    Forall((A_1_to_n, x),
           XOr(ExprRange(i, InSet(x, IndexedVar(A, i)), one, n)),
    conditions=[InSet(x, SymmetricDifference(A_1_to_n))]),
    domain=NaturalPos)
)
```

```python
membership_folding = (
    Forall(n,
    Forall((A_1_to_n, x),
           InSet(x, SymmetricDifference(A_1_to_n)),
    conditions=[XOr(ExprRange(i, InSet(x, IndexedVar(A, i)), one, n))]),
    domain=NaturalPos)
)
```

```python
sym_diff_as_union = (
    Forall((A, B), Equals(SymmetricDifference(A, B), Union(Difference(A, B), Difference(B, A))))
)
```

```python
sym_diff_as_diff = (
    Forall((A, B), Equals(SymmetricDifference(A, B), Difference(Union(A, B), Intersect(A, B))))
)
```

#### Special Edge Cases

```python
empty_sym_diff_eval = Equals(SymmetricDifference(), EmptySet)
```

```python
unary_sym_diff_reduction = Forall(A, Equals(SymmetricDifference(A), A))
```

#### Identities and Anihilation

```python
sym_diff_annihilation = (
    Forall(A, Equals(SymmetricDifference(A, A), EmptySet))
)
```

```python
sym_diff_with_empty_left = (
    Forall(A, Equals(SymmetricDifference(EmptySet, A), A))
)
```

```python
sym_diff_with_empty_right = (
    Forall(A, Equals(SymmetricDifference(A, EmptySet), A))
)
```

```python
sym_diff_with_empty = (
    Forall((m, n),
    Forall((A_1_to_m, B_1_to_n),
        Equals(SymmetricDifference(A_1_to_m, EmptySet, B_1_to_n),
               SymmetricDifference(A_1_to_m, B_1_to_n)).with_wrap_before_operator()
    ),
    domain=Natural)
)
```

```python
sym_diff_reduction = (
    Forall((l, m, n),
    Forall((A_1_to_l, B_1_to_m, C_1_to_n, X),
    Equals(SymmetricDifference(A_1_to_l, X, B_1_to_m, X, C_1_to_n),
           SymmetricDifference(A_1_to_l, B_1_to_m, C_1_to_n)).with_wrap_before_operator()
    ),
    domain=Natural)
)
```

#### Commutativity

```python
commutation = Forall((A, B), Equals(SymmetricDifference(A, B), SymmetricDifference(B, A)))
```

```python
rightward_commutation = (
    Forall((l, m, n),
    Forall((A_1_to_l, B, C_1_to_m, D_1_to_n),
          Equals(SymmetricDifference(A_1_to_l, B, C_1_to_m, D_1_to_n),
                 SymmetricDifference(A_1_to_l, C_1_to_m, B, D_1_to_n)).with_wrap_before_operator()
    ),
    domain=Natural)
)
```

```python
leftward_commutation = (
    Forall((l, m, n),
    Forall((A_1_to_l, B_1_to_m, C, D_1_to_n),
           Equals(SymmetricDifference(A_1_to_l, B_1_to_m, C, D_1_to_n),
                  SymmetricDifference(A_1_to_l, C, B_1_to_m, D_1_to_n)).with_wrap_before_operator()
          ),
          domain=Natural)
)
```

#### Associativity

```python
association_trinary = (
    Forall((A, B, C), Equals(SymmetricDifference(SymmetricDifference(A, B), C),
                             SymmetricDifference(A, SymmetricDifference(B, C))))
)
```

```python
association = (
    Forall((l, m, n),
    Forall((A_1_to_l, B_1_to_m, C_1_to_n),
           Equals(SymmetricDifference(A_1_to_l, B_1_to_m, C_1_to_n),
                  SymmetricDifference(A_1_to_l, SymmetricDifference(B_1_to_m), C_1_to_n)).with_wrap_before_operator()
    ).with_wrapping(),
    domain=Natural).with_wrapping()
)
```

```python
disassociation = (
    Forall((l, m, n),
    Forall((A_1_to_l, B_1_to_m, C_1_to_n),
           Equals(SymmetricDifference(A_1_to_l, SymmetricDifference(B_1_to_m), C_1_to_n),
                  SymmetricDifference(A_1_to_l, B_1_to_m, C_1_to_n)).with_wrap_before_operator()
    ).with_wrapping(),
    domain=Natural).with_wrapping()
)
```

#### Distributivity

```python
distribute_intersection = (
    Forall(m,
    Forall((A, B_1_to_m, C),
           Equals(Intersect(A, SymmetricDifference(B_1_to_m), C),
                  SymmetricDifference(ExprRange(i, Intersect(A, IndexedVar(B, i), C), one, m))).with_wrap_after_operator()
          ),
    domain=Natural)
)
```
