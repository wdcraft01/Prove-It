```python
import proveit
# Prepare this notebook for defining the axioms of a theory:
%axioms_notebook # Keep this at the top following 'import proveit'.

from proveit import i, n, x, A, B, ExprRange, IndexedVar
from proveit.core_expr_types import A_1_to_n
from proveit.logic import Equals, Forall, InSet, XOr
from proveit.logic.sets import SymmetricDifference
from proveit.numbers import one, NaturalPos
```

```python
%begin axioms
```

```python
sym_diff_def = (
    Forall(n,
    Forall((x, A_1_to_n),
           Equals(InSet(x, SymmetricDifference(A_1_to_n)),
           XOr(ExprRange(i, InSet(x, IndexedVar(A, i)), one, n)))
    ),
    domain=NaturalPos)
)
```
