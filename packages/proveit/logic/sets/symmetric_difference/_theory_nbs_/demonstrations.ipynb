#### Working out the instantiation of the `membership_existential_unfolding` theorem

```python
from proveit import F, f, i, m, n, X, ExprRange, Function, IndexedVar
from proveit.numbers import one, Add, Natural, LessEq
```

#### Working out the instantiation of the `membership_existential_unfolding` theorem

```python
from proveit.logic.sets.symmetric_difference import membership_existential_unfolding
membership_existential_unfolding
```

```python
example_expr_range = (
    ExprRange(i, Function(F, IndexedVar(f, i)), Add(m, one), n)
)
example_expr = InSet(x, SymmetricDifference(example_expr_range))
display(example_expr_range)
display(example_expr)
```

```python
assumptions = [InSet(m, Natural), InSet(n, Natural), LessEq(m, n)]
```

```python
_n_sub = example_expr.domain.operands.num_elements(assumptions=assumptions)
```

```python
membership_existential_unfolding.instantiate(
    {n:_n_sub}, num_forall_eliminations=1, assumptions=assumptions)
```

```python
# InSet(x, SymmetricDifference(A, B, C)).existential_unfold(assumptions=[InSet(x, SymmetricDifference(A, B, C))])
```

```python
from proveit.logic import XOr
InSet(x, SymmetricDifference(A, B, C)).conclude(assumptions=[XOr(InSet(x, A), InSet(x, B), InSet(x, C))])
```

```python
InSet(x, SymmetricDifference(A, B, C, D)).prove(
        assumptions=[XOr(InSet(x, A), InSet(x, B), InSet(x, C), InSet(x, D))])
```
