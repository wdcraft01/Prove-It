## Miscellaneous Testing, Demonstrations

The material below was developed to test various symmetric difference-related constructions and methods. Some of this material could be integrated into the `_demonstrations_` page eventually and/or deleted as development continues.

#### Basic Constructions & Displays

```python
SymmetricDifference(A, B, C)
```

```python
print(SymmetricDifference(A, B, C))
```

#### Basic Membership(s)

```python
InSet(x, SymmetricDifference(A, B, C))
```

```python
NotInSet(y, SymmetricDifference(A, B, C))
```

#### Methods

#### `commutation()`

```python
SymmetricDifference(A, B, C).commutation(0, 2)
```

#### `group_commutation()`

```python
from proveit import D, E
SymmetricDifference(A, B, C, D, E).group_commutation(1, 2, 2)
```

`permutation()`

```python
SymmetricDifference(A, B, C, D, E).permutation(new_order=[4, 3, 2, 1, 0])
```

#### `shallow_simplification()` (as accessed through `simplification()`)

```python
SymmetricDifference(A, B, C).simplification()
```

```python
SymmetricDifference().simplification()
```

```python
SymmetricDifference(A).simplification()
```

```python
SymmetricDifference(SymmetricDifference(A, B), C).simplification().without_wrapping()
```

```python
from proveit import D
SymmetricDifference(SymmetricDifference(A, SymmetricDifference(B, C)), D).simplification()
```

```python
SymmetricDifference(A, B, A, B, A).simplification().without_wrapping()
```

```python
SymmetricDifference(A, A).simplification().without_wrapping()
```
