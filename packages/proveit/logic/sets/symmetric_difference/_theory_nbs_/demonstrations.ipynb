##### The `membership_existential_unfolding` theorem

```python
from proveit.logic.sets.symmetric_difference import membership_existential_unfolding
membership_existential_unfolding
```

Here is a proposed alternative form of the theorem, utilizing a “SymmetricDifferenceAll” formulation (the infrastructure for which is not yet established):

$\forall_{n \in \mathbb{N}}\forall_{A}\forall_{x}\big[x \in \Delta_{i=1}^{n}\big[A(i)\big] \Rightarrow \exists_{i \in \{1..n\}}\big(x \in A(i)\big)\big]$

which would then utilize an instantiation replacement map for $A$ that involves a lambda map, looking something like this: $\{A : \text{\\} i \mapsto F(f_{i})\}$. Work is currently underway on this approach.

$\forall_{n \in \mathbb{N}}\forall_{A}\big[\Delta_{i=1}^{n}\big[A(i)\big] = \big[x\big]\big]$

```python
# 2 useful expressions for our experimentation
n_minus_m, j_minus_m = subtract(n, m), subtract(j, m)
```

```python
# Expressions and ExprRanges for our experimentation
# utilizing 1..(n-m) vs. 1..(j-m)
example_expr_range = ExprRange(i, Function(F, IndexedVar(f, Add(m, i))), one, n_minus_m)
example_expr_range_j = ExprRange(i, Function(F, IndexedVar(f, Add(m, i))), one, subtract(j, m))
example_expr = InSet(x, SymmetricDifference(example_expr_range))
example_expr_j = InSet(x, SymmetricDifference(example_expr_range_j))
A_expr_range = ExprRange(i, IndexedVar(A, i), one, n_minus_m)
A_j_expr_range = ExprRange(i, IndexedVar(A, i), one, j_minus_m)

display(example_expr_range)
display(example_expr_range_j)
display(example_expr)
display(example_expr_j)
display(A_expr_range)
display(A_j_expr_range)
```

#### Instantiating only $n$

In the next 2 cells we instantiate only the original $n$ for a single forall elimination.

The first instantiation illustrates an interesting error: notice the subscripting on the forall-quantified $A$s, ending with the surprising $A_{n - m - m}$ term. We can avoid the error by using $\{n:j-m\}$ instead.

```python
# original instantiation using {n : n - m}
membership_existential_unfolding.instantiate(
    {n:n_minus_m}, num_forall_eliminations=1, assumptions=assumptions_n).with_wrapping()
```

```python
membership_existential_unfolding.instantiate(
    {n:j_minus_m}, num_forall_eliminations=1, assumptions=assumptions_j).with_wrapping()
```

In instantiation attempts further below, we apparently need to establish some basic partitioning of the underlying `ExprRanges` $1, 2, \ldots, n-m$ and $1, 2, \ldots, j-m$, so we do that here in the next four cells:

```python
basic_partition_01 = ExprRange(i, i, one, n_minus_m).partition(i, assumptions=assumptions_n)
```

```python
basic_partition_02 = basic_partition_01.inner_expr().rhs[0].split(subtract(i, one), assumptions=assumptions_n)
```

```python
basic_partition_j_01 = ExprRange(i, i, one, j_minus_m).partition(i, assumptions=assumptions_j)
```

```python
basic_partition_j_02 = basic_partition_j_01.inner_expr().rhs[0].split(subtract(i, one), assumptions=assumptions_j)
```
