Theorems (or conjectures) for the theory of <a class="ProveItLink" href="theory.ipynb">proveit.logic.sets.intersection</a>
========

```python
import proveit
# Prepare this notebook for defining the theorems of a theory:
%theorems_notebook # Keep this at the top following 'import proveit'.
from proveit import m, n, x, A
from proveit.core_expr_types import A_1_to_m, B_1_to_n
from proveit.logic import Forall, And, Equals, InSet, NotInSet, Intersect
from proveit.logic.sets import EmptySet, x_in_every_A, x_notin_some_A
from proveit.numbers import NaturalPos

%begin theorems
```

```python
membership_unfolding = Forall(m, Forall((x, A_1_to_m), x_in_every_A,
                                       conditions=[InSet(x, Intersect(A_1_to_m))]),
                             domain=NaturalPos)
```

```python
membership_folding = Forall(m, Forall((x, A_1_to_m), InSet(x, Intersect(A_1_to_m)),
                                     conditions=[x_in_every_A]),
                           domain=NaturalPos)
```

```python
nonmembership_equiv = Forall(m, Forall((x, A_1_to_m), 
                                      Equals(NotInSet(x, Intersect(A_1_to_m)), 
                                             x_notin_some_A)),
                           domain=NaturalPos)
```

```python
nonmembership_folding = Forall(m, Forall((x, A_1_to_m), 
                                        NotInSet(x, Intersect(A_1_to_m)), 
                                        conditions=[x_notin_some_A]),
                              domain=NaturalPos)
```

##### Intersections with Empty Set $\emptyset$

```python
intersect_with_empty_left = Forall(A, Equals(Intersect(EmptySet, A), EmptySet))
```

```python
intersect_with_empty_right = Forall(A, Equals(Intersect(A, EmptySet), EmptySet))
```

```python
intersect_with_empty = (
    Forall((m, n),
    Forall((A_1_to_m, B_1_to_n),
        Equals(Intersect(A_1_to_m, EmptySet, B_1_to_n ), EmptySet)),
    domain=NaturalPos)
)
```
