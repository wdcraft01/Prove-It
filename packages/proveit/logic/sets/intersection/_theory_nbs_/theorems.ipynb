```python
import proveit
# Prepare this notebook for defining the theorems of a theory:
%theorems_notebook # Keep this at the top following 'import proveit'.
from proveit import (
        i, j, k, l, m, n, x, A, B, C, S, T, ExprRange, Function, IndexedVar)
from proveit.core_expr_types import (
        A_i, A_1_to_l, A_1_to_m, B_j, B_1_to_m, B_1_to_n,
        C_1_to_m, C_1_to_n, D_1_to_n)
from proveit.logic import And, Equals, Forall, Implies, InSet, NotEquals, NotInSet
from proveit.logic.sets import (
        Difference, EmptySet, Intersect, IntersectAll, Union, UnionAll,
        x_in_every_A, x_notin_some_A)
from proveit.numbers import (
        one, Add, Interval, Integer, Less, LessEq, Natural, NaturalPos)

%begin theorems
```

```python
membership_unfolding = Forall(m, Forall((x, A_1_to_m), x_in_every_A,
                                       conditions=[InSet(x, Intersect(A_1_to_m))]),
                             domain=NaturalPos)
```

##### Special “Edge” Cases

Notice that, in contrast to the situation for _unions_, the _nullary intersection_ or intersection over an empty collection, $[\cap]()$, is problematic and often rather controversial, with the standard conclusion that $[\cap]()$ is equal to the universal class (or set of all sets). For a straightforward presentation, consider the "Nullary Intersection" section of the _Wikipedia_ entry at https://en.wikipedia.org/wiki/Intersection_(set_theory), and for a taste of the often emotional controversy, one can sample some of the responses and arguments at https://math.stackexchange.com/questions/370188/empty-intersection-and-empty-union.

In standard Zermelo-Fraenkel (ZF) set theory, which lacks a universal set, the empty intersection is either considered undefined or is defined with respect to a predefined universe of discourse $X$ in which case $[\cap]() = \bigcap\emptyset =  X$.

##### Idempotency

```python
redundant_intersection_binary = Forall(A, Equals(Intersect(A, A), A))
```

```python
redundant_intersection_range = (
    Forall(n,
    Forall(A, Equals(Intersect(ExprRange(i, A, one, n)), A)),
    domain=NaturalPos)
)
```

```python
redundant_intersection_range_general = (
    Forall((i, j),
    Forall(A, Equals(Intersect(ExprRange(i, A, i, j)), A)),
    conditions=[Less(i, j)], domain=Integer)
)
```

##### Intersections with Empty Set $\emptyset$

```python
intersect_with_empty_left = Forall(A, Equals(Intersect(EmptySet, A), EmptySet))
```

```python
intersect_with_empty_right = Forall(A, Equals(Intersect(A, EmptySet), EmptySet))
```

```python
intersect_with_empty = (
    Forall((m, n),
    Forall((A_1_to_m, B_1_to_n),
        Equals(Intersect(A_1_to_m, EmptySet, B_1_to_n ), EmptySet)),
    domain=Natural)
)
```

##### Non-Empty Intersections

```python
intersect_nonemptiness = (
    Forall((A, B),
           Implies(NotEquals(Intersect(A, B), EmptySet),
                   And(NotEquals(A, EmptySet), NotEquals(B, EmptySet))))
)
```

##### Commutativity

```python
commutation = Forall((A, B), Equals(Intersect(A, B), Intersect(B, A)))
```

```python
rightward_commutation = (
    Forall((l, m, n),
    Forall((A_1_to_l, B, C_1_to_m, D_1_to_n),
           Equals(Intersect(A_1_to_l, B, C_1_to_m, D_1_to_n),
                  Intersect(A_1_to_l, C_1_to_m, B, D_1_to_n)).with_wrap_before_operator()
    ).with_wrapping(),
    domain=Natural).with_wrapping()
)
```

```python
leftward_commutation = (
    Forall((l, m, n),
    Forall((A_1_to_l, B_1_to_m, C, D_1_to_n),
           Equals(Intersect(A_1_to_l, B_1_to_m, C, D_1_to_n),
                  Intersect(A_1_to_l, C, B_1_to_m, D_1_to_n)).with_wrap_before_operator()
    ).with_wrapping(),
    domain=Natural).with_wrapping()
)
```

##### Associativity

```python
association_trinary = (
    Forall((A, B, C),
           Equals(Intersect(Intersect(A, B), C), Intersect(A, Intersect(B, C))))
)
```

```python
association = (
    Forall((l, m, n),
    Forall((A_1_to_l, B_1_to_m, C_1_to_n),
           Equals(Intersect(A_1_to_l, B_1_to_m, C_1_to_n),
                  Intersect(A_1_to_l, Intersect(B_1_to_m), C_1_to_n)).with_wrap_before_operator()
    ).with_wrapping(),
    domain=Natural).with_wrapping()
)
```

```python
disassociation = (
    Forall((l, m, n),
    Forall((A_1_to_l, B_1_to_m, C_1_to_n),
           Equals(Intersect(A_1_to_l, Intersect(B_1_to_m), C_1_to_n),
                  Intersect(A_1_to_l, B_1_to_m, C_1_to_n)).with_wrap_before_operator()
    ).with_wrapping(),
    domain=Natural).with_wrapping()
)
```

```python
intersect_eq_intersectall = (
    Forall(
    (i, j), Forall(A, Equals(Intersect(ExprRange(k, Function(A, k), i, j)),
                             IntersectAll(k, Function(A, k), domain=Interval(i,j)))),
    conditions=[LessEq(i, Add(j, one))],
    domain=Integer)
)
```

```python
distribute_through_union = (
    Forall((l, m, n),
        Forall((A_1_to_l, B_1_to_m, C_1_to_n),
            Equals(Intersect(A_1_to_l, Union(B_1_to_m), C_1_to_n),
                   Union(ExprRange(i,
                                       Intersect(A_1_to_l, IndexedVar(B, i), C_1_to_n),
                                       one, m)
                   )
            ).with_wrapping_at(2),
        ).with_wrapping(),
    domain=Natural).with_wrapping()
)
```

```python
distribute_through_unionall = (
    Forall((l, n),
        Forall((S, A_1_to_l, B, C_1_to_n),
            Equals(Intersect(A_1_to_l, UnionAll(i, Function(B, i), S), C_1_to_n),
                   UnionAll(i, Intersect(A_1_to_l, Function(B, i), C_1_to_n), S)
            ).with_wrapping_at(2),
        ).with_wrapping(),
    domain=Natural).with_wrapping()
)
```

```python
from proveit import D, E
from proveit.numbers import two
distribute_through_difference = (
    Forall((l, n),
        Forall((A_1_to_l, D, E, C_1_to_n),
            Equals(Intersect(A_1_to_l, Difference(D, E), C_1_to_n),
                   Difference(Intersect(A_1_to_l, D, C_1_to_n),
                              Intersect(A_1_to_l, E, C_1_to_n)
                   )
            ).with_wrapping_at(2),
        ).with_wrapping(),
    domain=Natural).with_wrapping()
)
```
