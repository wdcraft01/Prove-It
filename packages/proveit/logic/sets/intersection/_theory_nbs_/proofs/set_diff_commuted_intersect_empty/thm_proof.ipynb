```python
import proveit
theory = proveit.Theory() # the theorem's theory
from proveit import defaults, display_provers # useful imports
from proveit import x
from proveit.logic import Exists, InSet, NotEquals
from proveit.logic.sets import EmptySet
```

```python
%proving set_diff_commuted_intersect_empty
```

We try a proof by contradiction, beginning with the assumption that $(A-B) \cap (B-A) \ne \emptyset$.

```python
intersect_commuted_differences = set_diff_commuted_intersect_empty.instance_expr.lhs
```

```python
contr_assumption = NotEquals(intersect_commuted_differences, EmptySet)
```

```python
defaults.assumptions = [contr_assumption]
```

```python
exists_x_in_intersect = Exists(x, InSet(x, intersect_commuted_differences)).prove()
```

```python
from proveit import Variable
x_star = Variable("x*", r'x^{*}', fence_when_forced=True)
```

```python
exists_x_in_intersect.choose(x_star)
```

```python
# check the default assumptions
defaults.assumptions
```

```python
x_star_elem_assumption = (
    next((a for a in defaults.assumptions if (hasattr(a, 'element') and a.element==x_star)), None)
)
```

```python
x_star_in_both_diffs = x_star_elem_assumption.unfold()
```

```python
x_star_in_A_but_not_B = x_star_in_both_diffs.operands[0].unfold()
```

```python
x_star_in_B_but_not_A = x_star_in_both_diffs.operands[1].unfold()
```

```python
x_star_in_A = x_star_in_A_but_not_B.derive_left()
```

```python
x_star_not_in_A = x_star_in_B_but_not_A.derive_right()
```

```python
not_x_star_in_A = x_star_not_in_A.unfold()
```

```python
contradiction = not_x_star_in_A.derive_contradiction()
```
