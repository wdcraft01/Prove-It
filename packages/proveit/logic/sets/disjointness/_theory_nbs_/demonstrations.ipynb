### Testing the `Disjoint` class methods

The `Disjoint` class currently has just the `conclude()` method, which is specialized to deal only with cases of the form `Disjoint(A-B, B-A)` (such commuted set differences are always disjoint) or `Disjoint(A-B, Intersect(A,B))`, and cases in which one or more of the target sets itself has a `deduce_disjointness()` method.<br/>

```python
Disjoint(Difference(A, B), Difference(B, A)).prove()
```

```python
Disjoint(Difference(C, D), Difference(D, C)).prove()
```

```python
Disjoint(Difference(A, B), Intersect(A, B)).prove()
```
