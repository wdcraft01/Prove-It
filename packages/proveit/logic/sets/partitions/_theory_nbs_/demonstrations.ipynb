### Testing of Partition-related Classes & Methods

#### Basic Constructions

Partitions(S) denotes the set of all partitions of set $S$.

```python
Partitions(S)
```

We can claim that $P$ is a partition of set $S$ by claiming that $P$ is in `Partitions(S)`.

```python
InSet(P, Partitions(S))
```

#### Testing `PartitionsMembership` methods

```python
InSet(P, Partitions(S)).deduce_in_bool()
```

```python
InSet(P, Partitions(S)).definition()
```

```python
InSet(P, Partitions(S)).as_defined()
```

```python
in_partitions_means_in_setofall = InSet(P, Partitions(S)).unfold(assumptions = [InSet(P, Partitions(S))])
```

#### `PartitionsMembership().conclude()`

##### (1) `PartitionsMembership().conclude()` can deal with several different cases, from very general to quite specific. We can see the most general case in the following steps:

```python
temp_assumption = in_partitions_means_in_setofall.expr
```

```python
InSet(P, Partitions(S)).conclude(assumptions = [temp_assumption])
```

##### (2) At the other extreme of the spectrum of possibilities, `PartitionsMembership().conclude()` can also deal with the very specific trivial case of a non-empty set being partitioned by the singleton set containing itself:

```python
singleton_partitions_claim = InSet(Set(S), Partitions(S))
```

```python
singleton_partitions_claim.conclude(assumptions=[NotEquals(S, EmptySet)])
```

And such trivial cases include sets that consist of other sets:

```python
InSet(Set(Set(A, B)), Partitions(Set(A, B))).conclude(assumptions=[NotEquals(Set(A, B), EmptySet)])
```

##### (3) The bipartition case: $\{A, B\} \in \textrm{Partitions}(S)$

```python
from proveit import A, B
two_elem_set_partitions_claim = InSet(Set(A, B), Partitions(S))
```

```python
from proveit.logic import Union
two_elem_set_partitions_claim.conclude(
    assumptions=[NotEquals(A, EmptySet), NotEquals(B, EmptySet), Disjoint(A, B), Equals(Union(A, B), S)])
```

##### (4) The partition-by-subdivision case: $P \cup Q \in \textrm{Partitions}(S)$, where $P \cup Q$ produces a more fine-grained partition of $S$ compared to some other partition $\{A, B\}$ of $S$ we already know about (where $P \in \textrm{Partitions}(A)$ and $Q \in \textrm{Partitions}(B)$).

##### This process uses the following `partition_by_subdivision` theorem:

```python
from proveit.logic.sets.partitions import partition_by_subdivision
partition_by_subdivision
```

If we want to conclude that $(P \cup Q) \in \textrm{Partitions}(S)$ using the $\texttt{partitions\_by\_subdivision}$ theorem, we might try the following:

(1) Collect all memberships of the form $P \in \textrm{Partitions}(A)$ and extract the $A$s, generating a list $[A_{1}, \dots, A_{n}]$.

(2) Collect all memberships of the form $Q \in \textrm{Partitions}(B)$ and extract the $B$s, generating a list $[B_{1}, \ldots, B_{n}]$.

(3) We then have $[A_{1}, \dots, A_{n}]$ and $[B_{1}, \ldots, B_{n}]$ and want to know if there exists a set $\{A_{i}, B_{j}\} \in \textrm{Partitions}(S)$.

##### An example ensemble of known/assumed partition relations is shown below, and the `.conclude()` method result one cell further below:

```python
from proveit import A, B, C, D, P, Q, S
# various combinations of partitions are possible
defaults.assumptions = [
    InSet(P, Partitions(A)), InSet(P, Partitions(B)), InSet(Q, Partitions(C)), InSet(Q, Partitions(D)),
    InSet(Set(B, C), Partitions(S))]
```

```python
InSet(Union(P, Q), Partitions(S)).conclude()
```

#### Checking Side-Effects of `Partitions(S)` Membership

If $P \in \texttt{Partitions}(S)$, then we know that:
- $\bigcup\limits_{A \in P} A = S$
- $\forall_{A \in P}\big(A \ne \emptyset\big)$
- $\forall_{A, B \in P}\big(\texttt{disjoint}(A, B)\big)$


```python
defaults.assumptions = [InSet(P, Partitions(S))]
```

```python
InSet(P, SetOfAll(P, P,
                  conditions = [Equals(UnionAll(A, A, domain = P), S),
                                Forall(A, NotEquals(A, EmptySet), domain = P),
                                AllDisjoint(P)],
                  domain = PowerSet(PowerSet(S)))).prove()
```

```python
defaults.assumptions = []
```

#### Utilizing the Partitions(S) Concept?

At the time of this writing (August 2025), the Partitions concept is difficult to use at the level of proofs, because we still lack some infrastructure development to support the establishment of the required conditions that the elements of the supposed partition all consist of non-empty mutually-disjoint subsets of the original set whose union gives the original set.

For example, the sets $A = \{1, 2, 3, 4, 5\}$ and $B = \{6, 7, 8, 9, 10\}$ form a partition of the set $C = \{1, 2, \ldots, 10\}$. Can we prove that $\{A, B\} \in \texttt{Partitions}(C)$? Not at this time. It's difficult to even prove that $A$ and $B$ are each not equal to the empty set. And though we can pretty easily prove that $A$ and $B$ are disjoint, due to prior development of the enumerated set Set() class, it is then still difficult to convert that to a proof that $\forall_{A, B \in P}\big(\texttt{disjoint}(A, B)\big)$ for $P = \{A, B\}$.

Nevertheless, it could and will be useful to be able to posit or assume that some set of sets, say $P$, is a partition of some other set, say $S$, allowing us to then conclude that elements of $P$ are non-empty mutually-disjoint subsets of $S$ whose union gives $S$. The accessibility of such conclusions will depend on the development in the SetOfAll class.
