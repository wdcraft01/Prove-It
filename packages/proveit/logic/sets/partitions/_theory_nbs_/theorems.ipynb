```python
import proveit
# Prepare this notebook for defining the theorems of a theory:
%theorems_notebook # Keep this at the top following 'import proveit'.

from proveit import x, A, B, P, Q, S, X
from proveit.logic import And, Equals, Forall, in_bool, InSet, NotEquals
from proveit.logic.sets import (AllDisjoint, Disjoint, EmptySet, Partitions, PowerSet,
        SetOfAll, SubsetEq, UnionAll)
```

```python
%begin theorems
```

```python
partitions_membership_def = (
    Forall((S, X),
    Equals(InSet(X, Partitions(S)),
           InSet(X, SetOfAll(P, P,
               conditions = [Equals(UnionAll(A, A, domain = P), S),
                             Forall(A, NotEquals(A, EmptySet), domain = P),
                             AllDisjoint(P)],
               domain = PowerSet(PowerSet(S))))
    ))
)
```

```python
partitions_membership_unfolding = (
    Forall((S, X),
    InSet(X, SetOfAll(P, P,
               conditions = [Equals(UnionAll(A, A, domain = P), S),
                             Forall(A, NotEquals(A, EmptySet), domain = P),
                             AllDisjoint(P)],
               domain = PowerSet(PowerSet(S)))), conditions = [InSet(X, Partitions(S))])
)
```

```python
partitions_membership_folding = (
    Forall((S, X),
    InSet(X, Partitions(S)),
    conditions = [InSet(X, SetOfAll(P, P,
               conditions = [Equals(UnionAll(A, A, domain = P), S),
                             Forall(A, NotEquals(A, EmptySet), domain = P),
                             AllDisjoint(P)],
               domain = PowerSet(PowerSet(S))))])
)
```

```python
from proveit import k, n, A, X, IndexedVar
from proveit.core_expr_types import A_1_to_n
from proveit.logic import EmptySet, Implies, NotEquals, Set, Union
from proveit.numbers import one, Interval, NaturalPos
partitions_membership_folding_concrete = (
    Forall(S,
    Forall(n,
           Forall(A_1_to_n,
                  Implies(And(Equals(Union(A_1_to_n), S),
                              Forall(k, NotEquals(IndexedVar(A, k), EmptySet), domain=Interval(one, n)),
                              Disjoint(A_1_to_n)),
                          InSet(Set(A_1_to_n), Partitions(S))).with_wrap_before_operator()
           ).with_wrapping(),
    domain=NaturalPos).with_wrapping()
    ).with_wrapping()
)
```

```python
unipartition = Forall(S, InSet(Set(S), Partitions(S)), conditions=[NotEquals(S, EmptySet)])
```

```python
partition_by_subdivision = (
    Forall((A, B, S, P, Q),
    Implies(And(InSet(Set(A, B), Partitions(S)),
                InSet(P, Partitions(A)),
                InSet(Q, Partitions(B))),
            InSet(Union(P, Q), Partitions(S))).with_wrap_before_operator()
    )
)
```

```python
bipartition = (
    Forall(S, Forall((A, B),
                     Implies(And(Equals(Union(A, B), S),
                                 Disjoint(A, B)),
                             InSet(Set(A, B), Partitions(S))),
    conditions=[NotEquals(A, EmptySet), NotEquals(B, EmptySet)]))
)
```

```python
partition_elements_are_subsets = Forall((x, X, S), SubsetEq(x, S), conditions = [InSet(X, Partitions(S)), InSet(x, X)])
```

```python
partitions_membership_is_bool = (
    Forall((S, X), in_bool(InSet(X, Partitions(S))))
)
```
