```python
import proveit
# Prepare this notebook for defining the theorems of a theory:
%theorems_notebook # Keep this at the top following 'import proveit'.
from proveit import i, j, m, n, s, x, A, B, C, S, T, ExprRange, Function
from proveit.core_expr_types import A_i, A_1_to_m, B_j, B_1_to_n
from proveit.logic import (
        And, Boolean, Equals, Exists, Forall, InSet, NotInSet)
from proveit.logic.sets import (
        Card, EmptySet, Intersect, IntersectAll, SetEquiv, SubsetEq,
        Union, UnionAll)
from proveit.numbers import one, greater_eq, LessEq, Natural, NaturalPos

%begin theorems
```

**Nothing is in the empty set (everything is *not* in the empty set):**

```python
nothing_is_in_empty = Forall(x, NotInSet(x, EmptySet))
```

### Identities Involving Unions, Intersections, and Equivalences

```python
union_with_superset_is_superset = Forall(
        (A, S),
        SetEquiv(Union(A, S), S),
        conditions=[SubsetEq(A, S)])
```

```python
intersection_with_superset_is_set = Forall(
        (A, S),
        SetEquiv(Intersect(A, S), A),
        conditions=[SubsetEq(A, S)])
```

```python
set_subset_eq_of_union_with_set = Forall(
    (A, B),
    SubsetEq(A, Union(A, B)),
    condition=Forall(x, InSet(InSet(x, B), Boolean)))
```

```python
intersection_is_subset_eq = Forall(
        (A, B),
        SubsetEq(Intersect(A, B), A),
    condition=Forall(x, InSet(InSet(x, B), Boolean)))
```

```python
intersection_subset_eq_union = Forall(
        (A, B),
        SubsetEq(Intersect(A, B), Union(A, B)))
```

```python
distribution_intersection_over_union = (
    Forall((A, B, C),
    Equals(Intersect(A, Union(B, C)),
           Union(Intersect(A, B), Intersect(A, C))))
)
```

```python
distribution_union_over_intersection = (
    Forall((A, B, C),
    Equals(Union(A, Intersect(B, C)),
           Intersect(Union(A, B), Union(A, C))))
)
```

```python
intersection_of_unions = (
    Forall((m, n),
    Forall((A_1_to_m, B_1_to_n),
    Equals(Intersect(Union(A_1_to_m), Union(B_1_to_n)),
           Union(ExprRange(i, ExprRange(j, Intersect(A_i, B_j),
                 one, n), one, m))
    ).with_wrap_after_operator()).with_wrapping(),
    domain=NaturalPos).with_wrapping()
)
```

```python
intersection_of_unionalls = (
    Forall((A, B, S, T),
    Equals(
    Intersect(UnionAll(i, Function(A, i), S), UnionAll(j, Function(B, j), T)),
    UnionAll(i, UnionAll(j, Intersect(Function(A, i), Function(B, j)), T), S)
    )
    )
)
```

```python
union_of_intersections = (
    Forall((m, n),
    Forall((A_1_to_m, B_1_to_n),
    Equals(Union(Intersect(A_1_to_m), Intersect(B_1_to_n)),
           Intersect(ExprRange(i, ExprRange(j, Union(A_i, B_j),
                 one, n), one, m))
    ).with_wrap_after_operator()).with_wrapping(),
    domain=NaturalPos).with_wrapping()
)
```

```python
union_of_intersectalls = (
    Forall((A, B, S, T),
    Equals(
    Union(IntersectAll(i, Function(A, i), S), IntersectAll(j, Function(B, j), T)),
    IntersectAll(i, IntersectAll(j, Union(Function(A, i), Function(B, j)), T), S)
    )
    )
)
```

### Existence, Subsets, & Cardinality

```python
element_implies_non_zero_card = (
    Forall(S, greater_eq(Card(S), one),
    conditions = [Exists(x, InSet(x, S))])
)
```

```python
element_eq_non_zero_card = (
    Forall(x,
        Forall(S,
               Equals( Exists(x, InSet(x, S)),
                       greater_eq(Card(S), one)),
        )
    )
)
```

```python
from proveit import Q, S, X, Function
from proveit.logic.sets.comprehension import SetOfAll
element_with_property_eq_non_zero_card = (
    Forall(x,
        Forall(Q,
        Forall(S,
            Equals(
            Exists(x, Function(Q, x), domain = S),
            greater_eq(Card(SetOfAll(x, x, condition = Function(Q, x), domain =S)), one)
            )
        )
        )
    )
)
```

```python
card_finite_subset_lesseq_card_superset = Forall((A, B),
LessEq(Card(A), Card(B)),
conditions = [InSet(Card(A), Natural), SubsetEq(A, B)])
```

```python
card_subset_le_card_finite_superset = Forall((A, B),
      LessEq(Card(A), Card(B)),
      conditions = [SubsetEq(A, B), InSet(Card(B), Natural)])
```

```python
finite_subset_with_card_of_superset_is_superset = (
        Forall((A, B),
                Equals(A, B),
        conditions = [InSet(Card(A), Natural), SubsetEq(A, B), Equals(Card(A), Card(B))])
)
```

```python
maximal_size_subset_is_superset = (
Forall(n,
Forall((A, B),
        And(Equals(Card(B), n), Equals(A, B)),
conditions = [SubsetEq(A, B), Equals(Card(A), n), LessEq(Card(B), n)]),
domain = Natural)
)
```

### 2-Block Pseudo-Partitions of Sets

```python
from proveit.logic import And, Difference, Iff, XOr
bipartition_eq_set_differences = Forall((A, B, S),
Iff(And(Equals(S, Union(A, B)), Equals(Intersect(A, B), EmptySet)),
   And(Equals(Difference(S, A), B), Equals(Difference(S, B), A)))
)
```

```python
bipartition_membership_eq_set_differences = Forall((A, B, S),
Iff(And(SubsetEq(A, S), SubsetEq(B, S), Forall(x, XOr(InSet(x, A), InSet(x, B)), domain = S)),
   And(Equals(Difference(S, A), B), Equals(Difference(S, B), A)))
)
```
