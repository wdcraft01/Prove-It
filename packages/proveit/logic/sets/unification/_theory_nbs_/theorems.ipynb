Theorems (or conjectures) for the theory of <a class="ProveItLink" href="theory.ipynb">proveit.logic.sets.unification</a>
========

```python
import proveit
# Prepare this notebook for defining the theorems of a theory:
%theorems_notebook # Keep this at the top following 'import proveit'.
from proveit import (i, j, k, l, m, n, x, A, B, C, S, X,
                     ExprRange, Function, IndexedVar)
from proveit.core_expr_types import (
        A_1_to_l, A_1_to_m, B_1_to_m, B_1_to_n, C_1_to_m,
        C_1_to_n, D_1_to_n)
from proveit.logic import Forall, And, Or, Equals, InSet, NotInSet 
from proveit.logic.sets import (
        each_A_in_S, EmptySet, Intersect, IntersectAll, SubsetEq,
        Union, UnionAll, x_in_any_A, x_notin_all_A)
from proveit.numbers import (
        one, Add, Integer, Interval, Less, LessEq, Natural, NaturalPos)

%begin theorems
```

```python
membership_unfolding = Forall(m, Forall((x, A_1_to_m), x_in_any_A, 
                                       conditions=[InSet(x, Union(A_1_to_m))]),
                             domain=NaturalPos)
```

```python
membership_folding = Forall(m, Forall((x, A_1_to_m), InSet(x, Union(A_1_to_m)), 
                                     conditions=[x_in_any_A]),
                           domain=NaturalPos)
```

```python
nonmembership_equiv = Forall(m, Forall((x, A_1_to_m), 
                                      Equals(NotInSet(x, Union(A_1_to_m)),
                                             x_notin_all_A)),
                            domain=NaturalPos)
```

```python
nonmembership_folding = Forall(m, Forall((x, A_1_to_m), 
                                        NotInSet(x, Union(A_1_to_m)),
                                        conditions=[x_notin_all_A]),
                              domain=NaturalPos)
```

```python
union_inclusion = Forall(m, Forall((A_1_to_m, S), 
                                   SubsetEq(Union(A_1_to_m), S),
                                   conditions=[each_A_in_S]),
                         domain=NaturalPos)
```

##### Special “Edge” Cases

```python
empty_union_eval = Equals(Union(), EmptySet)
```

```python
unary_union_reduction = Forall(A, Equals(Union(A), A))
```

##### Idempotency

```python
redundant_union_binary = Forall(A, Equals(Union(A, A), A))
```

```python
redundant_union_range = (
    Forall(n,
    Forall(A, Equals(Union(ExprRange(i, A, one, n)), A)),
    domain=NaturalPos)
)
```

```python
redundant_union_range_general = (
    Forall((i, j),
    Forall(A, Equals(Union(ExprRange(i, A, i, j)), A)),
    conditions=[Less(i, j)], domain=Integer)
)
```

##### Unions with Empty Set $\emptyset$

```python
union_with_empty_left = Forall(A, Equals(Union(EmptySet, A), A))
```

```python
union_with_empty_right = Forall(A, Equals(Union(A, EmptySet), A))
```

```python
union_with_empty = (
    Forall((m, n),
    Forall((A_1_to_m, B_1_to_n),
        Equals(Union(A_1_to_m, EmptySet, B_1_to_n ), Union(A_1_to_m, B_1_to_n ))),
    domain=Natural)
)
```

##### Commutativity

```python
commutation = Forall((A, B), Equals(Union(A, B), Union(B, A)))
```

```python
rightward_commutation = (
    Forall((l, m, n),
    Forall((A_1_to_l, B, C_1_to_m, D_1_to_n),
           Equals(Union(A_1_to_l, B, C_1_to_m, D_1_to_n),
                  Union(A_1_to_l, C_1_to_m, B, D_1_to_n)).with_wrap_before_operator()
    ).with_wrapping(),
    domain=Natural).with_wrapping()
)
```

```python
leftward_commutation = (
    Forall((l, m, n),
    Forall((A_1_to_l, B_1_to_m, C, D_1_to_n),
           Equals(Union(A_1_to_l, B_1_to_m, C, D_1_to_n),
                  Union(A_1_to_l, C, B_1_to_m, D_1_to_n)).with_wrap_before_operator()
    ).with_wrapping(),
    domain=Natural).with_wrapping()
)
```

##### Associativity

```python
association_trinary = (
    Forall((A, B, C),
           Equals(Union(Union(A, B), C), Union(A, Union(B, C))))
)
```

```python
association = (
    Forall((l, m, n),
    Forall((A_1_to_l, B_1_to_m, C_1_to_n),
           Equals(Union(A_1_to_l, B_1_to_m, C_1_to_n),
                  Union(A_1_to_l, Union(B_1_to_m), C_1_to_n)).with_wrap_before_operator()
    ).with_wrapping(),
    domain=Natural).with_wrapping()
)
```

```python
disassociation = (
    Forall((l, m, n),
    Forall((A_1_to_l, B_1_to_m, C_1_to_n),
           Equals(Union(A_1_to_l, Union(B_1_to_m), C_1_to_n),
                  Union(A_1_to_l, B_1_to_m, C_1_to_n)).with_wrap_before_operator()
    ).with_wrapping(),
    domain=Natural).with_wrapping()
)
```

#### END WORK IN PROGRESS (ABOVE)

```python
union_eq_unionall = (
    Forall(
    (i, j), Forall(A, Equals(Union(ExprRange(k, Function(A, k), i, j)),
                             UnionAll(k, Function(A, k), domain=Interval(i,j)))),
    conditions=[LessEq(i, Add(j, one))],
    domain=Integer)
)
```

```python
distribute_through_intersection = (
    Forall((l, m, n),
        Forall((A_1_to_l, B_1_to_m, C_1_to_n),
            Equals(Union(A_1_to_l, Intersect(B_1_to_m), C_1_to_n),
                   Intersect(ExprRange(i,
                                       Union(A_1_to_l, IndexedVar(B, i), C_1_to_n),
                                       one, m)
                   )
            ).with_wrapping_at(2),
        ).with_wrapping(),
    domain=Natural).with_wrapping()
)
```

```python
distribute_through_intersectall = (
    Forall((l, n),
        Forall((S, A_1_to_l, B, C_1_to_n),
            Equals(Union(A_1_to_l, IntersectAll(i, Function(B, i), S), C_1_to_n),
                   IntersectAll(i, Union(A_1_to_l, Function(B, i), C_1_to_n), S)
            ).with_wrapping_at(2),
        ).with_wrapping(),
    domain=Natural).with_wrapping()
)
```
