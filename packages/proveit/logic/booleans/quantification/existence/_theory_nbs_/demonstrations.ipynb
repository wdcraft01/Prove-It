### Testing

#### Generate some existential test cases, each quite simple but varying across the collection in terms of the presence/absence of explicit conditions and domain(s).

```python
# some Exists() expressions
exists_test_exprs = [
    Exists((a, b), Equals(Add(a, b), five)), # no conditions at all
    Exists((a, b), Equals(Add(a, b), five), domain=NaturalPos), # single domain spec
    Exists((a, b), Equals(Add(a, b), five), domains=[NaturalPos, Integer]), # two domains
    Exists((a, b), Equals(Add(a, b), five), conditions = [LessEq(a, five), LessEq(b, five)]), # explicit conditions
    Exists((a, b), Equals(Add(a, b), five), conditions = [LessEq(a, five), LessEq(b, five)], domain = Integer), # conditions + domain
    Exists((a, b, c), LessEq(Add(a, b, c), five), conditions = [greater_eq(a, one), greater_eq(b, one), greater_eq(c, one)], domain = Integer), # conditions + domain
    Exists(x, Px), # archetype case
    Exists((a, b), Not(Equals(Add(a, b), five))), # no conditions at all
    Exists((a, b), Not(Equals(Add(a, b), five)), domain = NaturalPos), # single domain spec
    Exists((a, b), Not(Equals(Add(a, b), five)), domains = [NaturalPos, Integer]), # two domains
    Exists((a, b), Not(Equals(Add(a, b), five)), conditions = [LessEq(a, five), LessEq(b, five)]), # explicit conditions
    Exists((a, b), Not(Equals(Add(a, b), five)), conditions = [LessEq(a, five), LessEq(b, five)], domain = Integer), # conditions + domain
    Exists(x, Not(Rx)) # archetype case
]
```

```python
# check to verify these look correct
for i, expr in enumerate(exists_test_exprs):
    display(i, expr)
```

```python
# add this as a default assumption so later judgments work out:
defaults.assumptions = [Exists(x, Px), Exists(x, Not(Rx))]
```

```python
# some Exists() judgments
exists_test_judgments = [
    exists_test_exprs[0].conclude_via_example((two, three)),
    exists_test_exprs[1].conclude_via_example((two, three)),
    exists_test_exprs[2].conclude_via_example((two, three)),
    exists_test_exprs[3].conclude_via_example((two, three)),
    exists_test_exprs[4].conclude_via_example((two, three)),
    exists_test_exprs[5].conclude_via_example((one, two, one)),
    exists_test_exprs[6].prove(), # proof by assumption
    exists_test_exprs[7].conclude_via_example((two, four)),
    exists_test_exprs[8].conclude_via_example((two, four)),
    exists_test_exprs[9].conclude_via_example((two, four)),
    exists_test_exprs[10].conclude_via_example((two, four)),
    exists_test_exprs[11].conclude_via_example((two, four)),
    exists_test_exprs[12].prove() # proof by assumption
]
```

```python
# check to verify the judgments look correct and as-expected:
for j in exists_test_judgments:
    display(j)
```

#### Now a test of the Exists().unfold() method

```python
for j in exists_test_judgments:
    display(j.unfold())
```

#### Testing the Exists().definition() method

```python
# we can do this on the exprs or the judgments
for j in exists_test_exprs:
    display(j.definition())
```

#### Testing `Exists.choose()`

```python
from proveit import a, b, c, d, i, k, ExprRange, ExprTuple, IndexedVar
from proveit.numbers import zero, one, three, Integer, NaturalPos
```

```python
a_1_to_k, a_1_to_3, b_1_to_3 = (
    ExprRange(i, IndexedVar(a, i), one, k),
    ExprRange(i, IndexedVar(a, i), one, three),
    ExprRange(i, IndexedVar(b, i), one, three)
)
```

```python
# an existential with conditions
exists_01 = Exists(a_1_to_3, Equals(Add(a_1_to_3), zero), domain=Integer)
```

```python
exists_01.conclude_via_example((zero, zero, zero))
```

```python
# an existential, without conditions
exists_02 = Exists(a_1_to_3, Equals(Add(a_1_to_3), zero))
```

```python
from proveit.numbers import Neg
exists_02.conclude_via_example((Neg(one), zero, one))
```

For a general skolemization, amounting to a simple replacement of the variable(s), we can provide the Variable to use as the substitute/replacement:

```python
bs_in_z, sum_is_zero = exists_01.choose(b)
```

```python
# here we have no conditions to producec
sum_is_zero = exists_02.choose(b)
```

```python
# clean up the assumptions before looking at more examples
defaults.assumptions = []
```

We can instead use an actual replacement map to achieve the same thing. Use of the replacement map has not been actively developed at this point, though:

```python
bs_in_z, sum_is_zero = exists_01.choose(replacement_map = {a:b})
```

```python
sum_is_zero = exists_02.choose(replacement_map = {a:b})
```

```python
# clean up the assumptions before looking at more examples
defaults.assumptions = []
```

If no conditions are present, we can Skolemize directly to another ExprRange:

```python
exists_02.choose(b_1_to_3)
```

```python
# clean up the assumptions before looking at more examples
defaults.assumptions = []
```

When conditions are present, trying to Skolemize directly to another ExprRange will fail. This is because of some underlying challenge in getting the replacements to work in the conditions. Instead, we now get a warning to use the Variable approach:

```python
try:
    exists_01.choose(b_1_to_3)
    print(f"ERROR! Should not get this far!")
except ValueError as the_err:
    print(f"EXCEPTION: {the_err}")
```

It's also tricky to skolemize an ExprRange to an explicit tuple or ExprTuple when the `Exists` has conditions. This is also due to some underlying difficulties in handling the associated replacement for those conditions. Just offering the explicit tuple or ExprTuple prompts an error message and some advice about how to proceed:

```python
try:
    exists_01.choose((b, c, d))
    print(f"WARNING: should not get this far!")
except ValueError as the_err:
    print(f"EXCEPTION: {the_err}")
```

As the instructions indicate, we need to supply not just the explicit tuple but also more explicit information about how to effect the replacement(s) in the condition(s). We can use the original `Exists` expression to see the format of those conditions, and then provide a `conditions_map` in the choose() method:

```python
exists_01.conditions
```

```python
from proveit.logic import InSet
exists_01.choose(b, c, d,
        conditions_map = {exists_01.conditions: (InSet(b, Integer), InSet(c, Integer), InSet(d, Integer))})
```

```python
# clean up the assumptions before looking at more examples
defaults.assumptions = []
```

Notice that when we have no conditions to worry about, we can Skolemize an ExprRange directly to an explicit tuple without any extra finagling:

```python
from proveit.logic import InSet
exists_02.choose(b, c, d)
```

```python
# clean up the assumptions before looking at more examples
defaults.assumptions = []
```
