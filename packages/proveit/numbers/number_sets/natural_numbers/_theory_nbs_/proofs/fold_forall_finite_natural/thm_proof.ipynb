========

```python
import proveit
theory = proveit.Theory() # the theorem's theory
from proveit import defaults, display_provers # useful imports
from proveit import m, n, N, P, Lambda
from proveit.logic import Equals, Forall, Implies, InSet
from proveit.numbers import zero, one, Add, LessEq, Natural, Neg, Pn
from proveit.numbers.number_sets.natural_numbers import fold_forall_natural
```

```python
%proving fold_forall_finite_natural
```

```python
defaults.assumptions = fold_forall_finite_natural.all_conditions()
```

```python
P_of_zero = fold_forall_finite_natural.instance_expr.instance_expr.antecedent.operands[0]
```

```python
P_of_m_plus_one = fold_forall_finite_natural.instance_expr.instance_expr.antecedent.operands[1]
```

```python
defaults.assumptions = defaults.assumptions + [P_of_zero, P_of_m_plus_one]
```

We capture the consequent of the theorem in the following definition of $Q(n)$:

```python
Q = Lambda(n, Implies(LessEq(n, N), Pn))
```

We then use our standard induction to prove $\forall_{n \in \mathbb{N}} [Q(n)]$.

Notice that $Q(0)$ is True (since we have $P(0)$ by assumption, $Q(0)$ is vacuously True):

```python
Q.apply(zero).prove()
```

Now we want to show that $Q(m) \Rightarrow Q(m+1)$.</br>
That is: we want to show that _if_ $[(m \le N) \Rightarrow P(m)]$, _then_ $(m+1 \le N) \Rightarrow P(m+1)$.

So we define $Q(m)$ and add $Q(m)$ to the assumptions:

```python
Q_of_m = Q.apply(m)
```

```python
defaults.assumptions = defaults.assumptions + [Q_of_m]
```

We want to prove that $Q(m+1)$ holds, _i.e._, that $(m+1 \le N) \Rightarrow P(m+1)$.<br/>
So we also now assume that $(m+1 \le N)$:

```python
m_plus_1_le_N = LessEq(Add(m, one), N)
```

```python
defaults.assumptions = defaults.assumptions + [m_plus_1_le_N]
```

We will also need to explicitly know that $m \in \mathbb{N}$:

```python
m_in_natural = InSet(m, Natural)
```

```python
defaults.assumptions = defaults.assumptions + [m_in_natural]
```

If $m+1 \le N$, then $m \le N$:

```python
m_plus_1_le_N_jdgmnt = m_plus_1_le_N.prove()
```

```python
# This is a bizarre and incorrect result; not sure what's happening
# Kept here for future reworking
# m_plus_1_le_N_jdgmnt.add_left(Neg(one))
```

```python
# So we work around it a bit, trying this first instead
m_plus_1_le_N_jdgmnt.add_left(Neg(one), strong=True)
```

```python
Q_of_m.derive_consequent()
```

```python
# Recall our quantified assumption
P_of_m_plus_one
```

```python
P_of_m_plus_one_jdgmnt = P_of_m_plus_one.instantiate({m:m})
```

```python
Q_of_m_plus_one_judgmnt = P_of_m_plus_one_jdgmnt.as_implication(m_plus_1_le_N)
```

```python
Q_of_m_plus_one_judgmnt.as_implication(Q_of_m)
```

To summarize thus far, we have shown:<br/>
(1) $Q(0)$ is True.</br>
(2) $Q(m) \Rightarrow Q(m+1)$ for $m \in \mathbb{N}$.

```python
fold_forall_natural_inst = fold_forall_natural.instantiate({P:Q})
```

Prove-It is not automatically recognizing that our quantified consequent above is what we need, so we prompt it manually to derive the equality of that consequent with the version in which we shift the $(n \le N)$ to a condition:

```python
eq_quantifiers = Equals(fold_forall_natural_inst.consequent,
                        Forall(n, Pn, domain=Natural, conditions=[LessEq(n, N)])).prove()
```
