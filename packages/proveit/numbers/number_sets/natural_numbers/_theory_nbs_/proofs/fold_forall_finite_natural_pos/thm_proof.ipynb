```python
import proveit
theory = proveit.Theory() # the theorem's theory
from proveit import defaults, display_provers # useful imports
```

```python
%proving fold_forall_finite_natural_pos
```

```python
defaults.assumptions = fold_forall_finite_natural_pos.all_conditions()
```

```python
P_of_one = fold_forall_finite_natural_pos.instance_expr.instance_expr.antecedent.operands[0]
```

```python
P_of_m_plus_one = fold_forall_finite_natural_pos.instance_expr.instance_expr.antecedent.operands[1]
```

```python
from proveit.numbers import Pone
defaults.assumptions += [P_of_one, P_of_m_plus_one]
defaults.assumptions
```

We will (eventually) use the standard (infinite) induction theorem to prove our finite case.

First, we define $Q(n)$ as follows:

```python
from proveit import n, N, Lambda
from proveit.logic import Implies
from proveit.numbers import LessEq
from proveit.numbers import Pn
Q = Lambda(n, Implies(LessEq(n, N), Pn))
```

Notice that $Q(1)$ is True (since we have $P(1)$ by assumption, $Q(1)$ is vacuously True):

```python
from proveit.numbers import one
Q.apply(one).prove()
```

Now we want to show that $Q(m) \Rightarrow Q(m+1)$.</br>
That is: we want to show that _if_ $[(m \le N) \Rightarrow P(m)]$, _then_ $(m+1 \le N) \Rightarrow P(m+1)$.

So we define $Q(m)$ and add $Q(m)$ to the assumptions:

```python
from proveit import m
Q_of_m = Q.apply(m)
```

```python
defaults.assumptions = defaults.assumptions + [Q_of_m]
```

We want to prove that $Q(m+1)$ holds, _i.e._, that $(m+1 \le N) \Rightarrow P(m+1)$.<br/>
So we also now assume that $(m+1 \le N)$:

```python
from proveit.numbers import one, Add, LessEq
m_plus_1_le_N = LessEq(Add(m, one), N)
```

```python
defaults.assumptions = defaults.assumptions + [m_plus_1_le_N]
```

We will also need to explicitly know that $m \in \mathbb{N}$:

```python
from proveit.logic import InSet
from proveit.numbers import NaturalPos
m_in_natural_pos = InSet(m, NaturalPos)
```

```python
defaults.assumptions = defaults.assumptions + [m_in_natural_pos]
```

If $m+1 \le N$, then $m \le N$:

```python
m_plus_1_le_N_jdgmnt = m_plus_1_le_N.prove()
```

```python
from proveit.numbers import Neg
# This is a bizarre and incorrect result; not sure what's happening
m_plus_1_le_N_jdgmnt.add_left(Neg(one))
```

```python
# So we work around it a bit, trying this first instead
m_plus_1_le_N_jdgmnt.add_left(Neg(one), strong=True)
```

```python
Q_of_m.derive_consequent()
```

```python
# Recall our quantified assumption
P_of_m_plus_one
```

```python
P_of_m_plus_one_jdgmnt = P_of_m_plus_one.instantiate({m:m})
```

```python
Q_of_m_plus_one_judgmnt = P_of_m_plus_one_jdgmnt.as_implication(m_plus_1_le_N)
```

```python
Q_of_m_plus_one_judgmnt.as_implication(Q_of_m)
```

To summarize thus far, we have shown:<br/>
(1) $Q(1)$ is True.</br>
(2) $Q(m) \Rightarrow Q(m+1)$ for $m \in \mathbb{N}^{+}$.

```python
from proveit.numbers.number_sets.natural_numbers import fold_forall_natural_pos
fold_forall_natural_pos
```

```python
from proveit import P
fold_forall_natural_pos_inst = fold_forall_natural_pos.instantiate({P:Q})
```

Prove-It is not automatically recognizing that our quantified consequent above is what we need, so we prompt it manually to derive the equality of that consequent with the version in which we shift the $(n \le N)$ to a condition:

```python
from proveit.logic import Equals, Forall
eq_quantifiers = Equals(fold_forall_natural_pos_inst.consequent,
                        Forall(n, Pn, domain=NaturalPos, conditions=[LessEq(n, N)])).prove()
```
