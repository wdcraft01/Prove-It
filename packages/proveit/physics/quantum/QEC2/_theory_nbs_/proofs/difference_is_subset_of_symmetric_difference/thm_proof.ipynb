```python
import proveit
theory = proveit.Theory() # the theorem's theory
from proveit import defaults, display_provers, A, B # useful imports
from proveit.logic.sets import SubsetEq, Difference
from proveit.logic.sets.symmetric_difference import sym_diff_as_union
```

```python
%proving difference_is_subset_of_symmetric_difference
```

```python
defaults.assumptions = difference_is_subset_of_symmetric_difference.all_conditions()
```

```python
A_delta_B_as_union = sym_diff_as_union.instantiate({A:A, B:B})
```

```python
A_diff_B_subset_union = SubsetEq(Difference(A, B), A_delta_B_as_union.rhs).prove()
```

```python
A_delta_B_as_union.sub_left_side_into(A_diff_B_subset_union)
```
