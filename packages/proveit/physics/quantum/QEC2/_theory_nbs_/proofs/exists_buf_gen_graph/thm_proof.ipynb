At this point, we know that $b$ consists of a set of $n$ faults, at least one of which (let's call it $f_{1}$) anti-commutes with logical operator $\ell$.

That fault $f_{1}$ can serve as a “base case” in constructing the desired fault sequence $(f_{1}, f_{2}, \ldots, f_{n})$ that serves as the desired “realization” of the $G_{S_{\ell}}$-path $p$.

Now, assume that we've partially constructed the desired fault sequence up through $(f_{1}, f_{2}, \ldots, f_{k})$ for some $k < n$, can we show that there exists some $f_{k+1}$ in the remaining faults making up $b$ so that:

$$H(\{f_{1}, f_{2}, \ldots, f_{k}\}) \cap H(\{f_{k+1}\}) \ne \emptyset$$

and

$$A_{\ell}(\{f_{1}, f_{2}, \ldots, f_{k}, f_{k+1}\}) \ne 1$$

unless $k = n-1$ ?

Now, assume that we've partially constructed the desired fault sequence up through $(f_{1}, f_{2}, \ldots, f_{k})$ for some $k < n$, can we show that for ANY choice of element in $H(\{f_{1}, f_{2}, \ldots, f_{k}\})$, there exists a fault among the remaining faults that will eliminate that detector?

YES!

Given the BUFILO $b = \{f_{1}, f_{2}, \ldots, f_{n}\}$. Then $H(b) = \emptyset$.

That means that any bipartition of the set of faults $b = A U B$, we have $H(A) = H(B)$. Thus, for example, in particular, $H(\{f_{1}, \ldots, f_{k}\}) = H(b - \{f_{1}, \ldots, f_{k}\})$.

$H(\{f_{1}\}) = \{a, b\}$ and $H(\{f_{2}\}) = \{b, c\}$ and $H(\{f_{3}\}) = \{c, d\}$, then $H(\{f_{1}, f_{2}, f_{3}\}) = \{a, d\}$

Special case: if current $A_{\ell}(\{f_{1}, f_{2}, \ldots, f_{k}\}) = 0$, there exists a fault $f^{*}$ among the remaining faults such that $A_{\ell}({f^{*}}) = 1$
