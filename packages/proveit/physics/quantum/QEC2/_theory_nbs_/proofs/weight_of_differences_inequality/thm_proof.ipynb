```python
import proveit
theory = proveit.Theory() # the theorem's theory
from proveit import defaults, display_provers, A, B, Lambda, x # useful imports
from proveit.numbers import LessEq, Neg
from proveit.logic.sets import Disjoint, Intersect, union_of_diff_and_intersect
from proveit.physics.quantum.QEC2 import Weight
```

```python
%proving weight_of_differences_inequality
```

```python
defaults.assumptions = weight_of_differences_inequality.all_conditions()
```

```python
A_as_union_of_diff_and_int = union_of_diff_and_intersect.instantiate({A:A, B:B})
```

```python
B_as_union_of_diff_and_int = union_of_diff_and_intersect.instantiate({A:B, B:A})
```

```python
Disjoint(*A_as_union_of_diff_and_int.rhs.operands).prove()
```

```python
Disjoint(*B_as_union_of_diff_and_int.rhs.operands).prove()
```

```python
weight_A_eq_weight_union = A_as_union_of_diff_and_int.substitution(Lambda(x, Weight(x)))
```

```python
weight_A_eq_weight_sum = weight_A_eq_weight_union.inner_expr().rhs.distribute_over_union()
```

```python
weight_B_eq_weight_union = B_as_union_of_diff_and_int.substitution(Lambda(x, Weight(x)))
```

```python
weight_B_eq_weight_sum = weight_B_eq_weight_union.inner_expr().rhs.distribute_over_union()
```

```python
inequality = defaults.assumptions[0]
```

```python
inequality = weight_A_eq_weight_sum.sub_right_side_into(inequality)
```

```python
inequality = weight_B_eq_weight_sum.sub_right_side_into(inequality)
```

```python
inequality = inequality.inner_expr().lhs.operands[1].operand.commute(0,1)
```
