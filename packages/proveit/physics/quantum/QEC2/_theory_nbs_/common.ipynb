```python
import proveit
# Prepare this notebook for defining the common expressions of a theory:
%common_expressions_notebook # Keep this at the top following 'import proveit'.

from proveit import Literal, Variable
from proveit.physics.quantum.QEC2 import (
        BufiloSetsLiteral, BufiloSequencesLiteral, DetectorsLiteral,
        ErrorsLiteral, FaultsLiteral, MalignantSetsLiteral, StatesLiteral, SyndromesLiteral)
```

```python
States = StatesLiteral()
```

##### A Maximum Weight $w_{\text{BUF}}$ for BUFILOs

```python
_max_buf_weight = Literal('w_BUF', r'w_{\textsc{buf}}')
```

##### A Literal $\ell$ to Represent the Logical Operator With Which the BUFILOs Anti-Commute

```python
_ell = Literal('l', r'\ell')
```

A literal $\nu$ to represent the choice function used for selecting a detector to de-activate.

```python
_nu = Literal('v', r'\nu', fence_when_forced=True)
```
