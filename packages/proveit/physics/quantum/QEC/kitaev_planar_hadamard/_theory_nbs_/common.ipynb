#### Boundary Points, Interior Points
<img src="../Kitaev_SC_points.png" width="300" height="300" align="right">
Some convenient expressions for referring to the sets of points marking the rough/smooth boundaries of the surface code, and the set of "interior" points (points not on the rough boundaries).</br>

The rough boundary points, abbreviated with the 'rb' notation, consists of two discontiguous sets $A$ and $B$, each of which is partitioned into two subsets for purposes of indexing in a set comprehension expression. See the image at right, where buondary sets $A$ and $B$ are shown in green and red, respectively, for the $i$th stage of boundary-transforming portion of the logical Hadamard process.

First, the rough boundary $A$ set (green points in the illustration above), as the union of two separately-indexed sets. These specifications for rough boundary $A$ work for $i \in \{1, 2, \ldots, n-1\}$ (when $i = n - 1$, we have a single point along the East in the `_rba2` set).

_Note: a formatting error currently keeps the ExprTuples from appearing as tuples, but the graph points are indeed being encoded as ordered pairs in a Prove-It ExprTuple._

```python
# East
_rba1 = SetOfAll(j, ExprTuple(_n, j), domain = Interval(i, subtract(_n, one)))
```

```python
# North
_rba2 = SetOfAll(k, ExprTuple(k, _n), domain = Interval(subtract(_n, i), subtract(_n, one)))
```

```python
# North & East
_rba = Union(_rba1, _rba2)
```

Then the rough boundary $B$ set (red points in the illustration above), again as the union of two separately-indexed sets. Unlike the case for rough boundary $A$, these specifications for rough boundary $B$ work correctly only for $i \in \{1, 2, \ldots, n-2\}$, with the $i=n-1$ case needing a separate specification:

```python
# West
from proveit.numbers import Add, Neg
_rbb1 = SetOfAll(ell, ExprTuple(zero, ell), domain = Interval(one, Add(_n, Neg(one), Neg(i))))
```

```python
# South
_rbb2 = SetOfAll(m, ExprTuple(m, zero), domain = Interval(one, i))
```

```python
# South & West
_rbb = Union(_rbb1, _rbb2)
```

For the rough boundary $B$ set at step $i=n-1$, we have only the south points:

```python
# South points for i = n - 1
_rbb_n_minus_1 = SetOfAll(m, ExprTuple(m, zero), domain = Interval(zero, subtract(_n, one)))
```

#### Interior Points (Non-Rough-Boundary Points)

```python
# old; see below for slightly different format
from proveit.logic.sets import Difference, Set
_interior_pts = (
    Difference(
        Difference(
            Difference(SetOfAll((j,k), ExprTuple(j,k), domain=Interval(zero, subtract(_n, one))),
                       Set(ExprTuple(zero, zero))),
            _rba),
        _rbb)
)
```

The definition below for interior points might work better, using the difference between a “core” set $\{j, k\}_{j, k \in \{0, \ldots, n-1\}}$ and the _union_ of the (rough) boundary points (and the origin), in part because it then becomes much easier to match the elements to an import disjointness-related theorem having the same format.

The first definition is for Hadamard pseudo-rotation index $i \in \{1, \ldots, n-2\}$. Then a separate second definition is needed for the case $i = n - 1$. Later we can fold in the cases for $i=0$ and $i=n$ (_i.e._, the beginning and end states).

```python
_interior_pts_i_1_to_n_minus_2 = Difference(SetOfAll((j,k), ExprTuple(j,k), domain=Interval(zero, subtract(_n, one))),
           Union(Set(ExprTuple(zero, zero)), _rba, _rbb))
```
