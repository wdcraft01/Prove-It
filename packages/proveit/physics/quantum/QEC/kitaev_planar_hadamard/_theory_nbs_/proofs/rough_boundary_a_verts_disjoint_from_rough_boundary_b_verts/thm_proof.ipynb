##### WORKED UP TO HERE

The challenge now is that we need to consider three cases for the value of $t' \in \{0, 1, \ldots, n-1, n\}$, which then determines the possible values of the Skolem constants $j' \in \{t,\ldots, n-1\} = \{t', \ldots, n-1\}$ and $m' \in \{1 - \delta_{t',(n-1)} - \delta_{t',n}, \ldots, t'\}$:

<b>Case 1:</b> $t' < n - 1$. This means that $\delta_{t',(n-1)} = \delta_{t',n} = 0$ and thus $m' \in \{1, \ldots, n-2\}$. Thus $n \ne m'$, contradicting the claim that $n=m'$.

<b>Case 2:</b> $t' = n - 1$. This means that $\delta_{t',(n-1)} = 1$ while $\delta_{t',n} = 0$ and thus $m' \in \{0, \ldots, n-1\}$. Again, $n \ne m'$, contradicting the claim that $n = m'$.

<b>Case 3:</b> $t' = n$. This means that $j' \in \{n, \ldots, n-1\} = \emptyset$, an impossibility and clearly contradicting the claim that $j' = 0$.
