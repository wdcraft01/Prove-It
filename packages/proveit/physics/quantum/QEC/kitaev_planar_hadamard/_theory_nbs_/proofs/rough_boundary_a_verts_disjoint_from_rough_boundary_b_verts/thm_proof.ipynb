Convert the `RBBVerts` to use $t'$ to avoid conflicts:

```python
rbb_verts_t_prime = rough_boundary_b_vertices_def.inner_expr().rhs.operand.relabeled({t:t_prime})
```

```python
# For convenience for later, name the four intervals:
j_set, k_set, ell_set, m_set = (
    rough_boundary_a_vertices_def.rhs.operand.body.value.operands[0].domain,
    rough_boundary_a_vertices_def.rhs.operand.body.value.operands[1].domain,
    rbb_verts_t_prime.rhs.operand.body.value.operands[0].domain,
    rbb_verts_t_prime.rhs.operand.body.value.operands[1].domain
)
```

```python
from proveit.physics.quantum.QEC.kitaev_planar_hadamard import (
        RoughBoundaryAVertices, RoughBoundaryBVertices)
current_eq = rough_boundary_a_vertices_def.expr.substitution(Intersect(RoughBoundaryAVertices, RoughBoundaryBVertices))
```

```python
current_eq = current_eq.inner_expr().rhs.operands[1].substitute(rbb_verts_t_prime)
```

```python
current_eq = current_eq.inner_expr().rhs.distribute(idx=0).inner_expr().rhs.operand.relabeled({i:t})
```

```python
current_eq = current_eq.inner_expr().rhs.instance_expr.distribute(1).inner_expr().rhs.instance_expr.operand.relabeled({i:t_prime})
```

```python
current_eq = current_eq.inner_expr().rhs.instance_expr.instance_expr.distribute(1)
```

```python
current_eq = current_eq.inner_expr().rhs.instance_expr.instance_expr.operands[0].distribute(0)
```

```python
current_eq = current_eq.inner_expr().rhs.instance_expr.instance_expr.operands[2].distribute(0)
```

`current_eq` has expressed the $\textrm{RBAVerts} \cap \textrm{RBBVerts}$ intersection as the union of four different intersections. 

We look at each of those four intersections separately, with the goal of showing that each intersection is actual the empty set $\emptyset$:

```python
intersection_01, intersection_02, intersection_03, intersection_04 = (
    current_eq.rhs.instance_expr.instance_expr.operands[0],
    current_eq.rhs.instance_expr.instance_expr.operands[1],
    current_eq.rhs.instance_expr.instance_expr.operands[2],
    current_eq.rhs.instance_expr.instance_expr.operands[3])
```
