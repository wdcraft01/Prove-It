#### General Plan

We can first show $\textrm{Disjoint}(\textrm{CoreVerts}, \textrm{RBAVerts})$, then conclude that $\textrm{Disjoint}(\textrm{XSiteVerts}, \textrm{RBAVerts})$, since $\textrm{XSiteVerts} = (\textrm{CoreVerts} - (\textrm{OriginVerts} \cup \textrm{RBBVerts}) \subseteq \textrm{CoreVerts}$.


Convert the `RBAVerts` to use $t'$ instead of $t$ to avoid conflicts:

```python
rba_verts_t_prime = rough_boundary_a_vertices_def.inner_expr().rhs.operand.relabeled({t:t_prime})
```

```python
from proveit.logic.sets import Intersect
from proveit.physics.quantum.QEC.kitaev_planar_hadamard import (
        CoreVertices, RoughBoundaryAVertices)
current_eq = core_vertices_def.substitution(Intersect(CoreVertices, RoughBoundaryAVertices))
```

```python
current_eq = current_eq.inner_expr().rhs.operands[1].substitute(rba_verts_t_prime)
```

```python
current_eq = current_eq.inner_expr().rhs.distribute(idx=0).inner_expr().rhs.operand.relabeled({i:t})
```

```python
current_eq = current_eq.inner_expr().rhs.instance_expr.distribute(idx=1).inner_expr().rhs.instance_expr.operand.relabeled({i:t_prime})
```

```python
current_eq = current_eq.inner_expr().rhs.instance_expr.instance_expr.distribute(idx=1)
```

`current_eq` has expressed the $\big[\textrm{CoreVerts} \cap \textrm{RBAVerts}\big]$ intersection as the union of two different intersections.<br/>
We now look at each of those two intersections separately, with the goal of showing that each intersection is actually the empty set $\emptyset$, and thus the entire expression is equal to the empty set $\emptyset$.

```python
# For convenience, label the intersection expressions
intersection_01, intersection_02 = (
    current_eq.rhs.instance_expr.instance_expr.operands[0],
    current_eq.rhs.instance_expr.instance_expr.operands[1])
```

```python
# For convenience, label the three set comprehension intervals
# and the two union-all intervals:
j_set, k_set, p_set, q_set, t_set, t_prime_set = (
    rba_verts_t_prime.rhs.operand.body.value.operands[0].domain,
    rba_verts_t_prime.rhs.operand.body.value.operands[1].domain,
    core_vertices_def.rhs.instance_expr.domain,
    core_vertices_def.rhs.instance_expr.domain,
    core_vertices_def.rhs.domain,
    rba_verts_t_prime.rhs.domain
)
```

```python
# For convenience, label the t, t' membership expressions
t_membership, t_prime_membership = InSet(t, t_set), InSet(t_prime, t_prime_set)
```
