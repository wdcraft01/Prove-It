```python
from proveit import t, ExprTuple
from proveit.numbers import one
selected_tuple = ExprTuple(one, one, t)
```

```python
core_t = CoreVertices.definition().rhs.instance_expr
```

##### Let's prove that $(1, 1, t) \in \{(p, q, t) | p,q\in\{0, \ldots, n-1\}\}$

```python
from proveit.logic import InSet
in_core_t_set_def = InSet(selected_tuple, core_t).definition()
```

##### We need a little more information about $1$ and the domain $\{0, \ldots, n-1\}$.

```python
from proveit.physics.quantum.QEC.kitaev_planar_hadamard import _n_ge_three, _n_in_integer
display(_n_ge_three)
display(_n_in_integer)
```

```python
InSet(one, in_core_t_set_def.rhs.domain).prove()
```

```python
exists_pq_judgment = in_core_t_set_def.rhs.conclude_via_example((one, one))
```

```python
in_core_t_set_def.sub_left_side_into(exists_pq_judgment)
```
