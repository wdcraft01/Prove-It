To show that the `UnionAll` is non-empty, we need to show that its “contents” are non-empty for at least one value of $t \in \{0, \ldots, n\}$. Our basic approach here is to show that (1) $\{(n, j, t) | j \in \{t, \ldots, n-1\}\} \ne \emptyset$ when we use $t = 0$; then (2) show that the extended union $\{(n, j, t) | j \in \{t, \ldots, n-1\}\} \cup \{(k, n, t) | k \in \{n-t, \ldots, n-1\}\} \ne \emptyset$ when we use $t=0$; and then conclude that the entire `UnionAll` is non-empty (since its “contents” are non-empty for at least one value in the `UnionAll`'s domain $\{0, \ldots, n\}$).

From inspection, we can see that the first component $\{(n, j, t) | j \in \{t, \ldots, n-1\}\}$ contains the 3-tuple $(n, 0, 0)$ and is thus non-empty when $t = 0$. We don't want to carry around an _assumption_ that $t = 0$, however (that makes things difficult); instead, we consider the specific subset $\{(n, j, 0) | j \in \{0, \ldots, n-1\}\}$.

```python
union_all = RoughBoundaryAVertices.definition().rhs
```

```python
RoughBoundaryAVertices.definition().rhs.instance_expr.operands[0]
```

```python
from proveit import j, ExprTuple
from proveit.logic.sets import SetOfAll
from proveit.numbers import zero, one, Interval, subtract
from proveit.physics.quantum.QEC.kitaev_planar_hadamard import _n
A_zero = SetOfAll(j, ExprTuple(_n, j, zero), domain=Interval(zero, subtract(_n, one)))
```

##### Now let's prove that $(n, 0, 0) \in A_{zero} = \{(n, j, 0)\}_{j \in \{0, \ldots, n-1\}}$. We do this via an existential in the `InSet` definition.

```python
selected_tuple = ExprTuple(_n, zero, zero)
```

```python
from proveit.logic.sets import InSet
inset_def = InSet(selected_tuple, A_zero).definition()
```

##### But we need a little more information about $0$ and the domain $\{0, \ldots, n-1\}$.

```python
from proveit.physics.quantum.QEC.kitaev_planar_hadamard import _n_ge_three, _n_in_integer
display(_n_ge_three)
display(_n_in_integer)
```

```python
InSet(zero, inset_def.rhs.domain).prove()
```

```python
exists_j_judgment = inset_def.rhs.conclude_via_example(zero)
```

```python
selected_tuple_in_set = inset_def.sub_left_side_into(exists_j_judgment)
```

```python
from proveit import x
from proveit.logic import Exists
something_exists_in_A_zero = Exists(x, InSet(x, A_zero))
```

```python
from proveit.logic.booleans.quantification.existence import existence_by_example_single
existence_by_example_single
```

```python
from proveit import P, Lambda
_P_sub = Lambda(x, something_exists_in_A_zero.instance_expr)
_x_sub = selected_tuple # we've now proven this is a valid example element
existence_by_example_single_inst = existence_by_example_single.instantiate(
            {P:_P_sub, x:_x_sub})
```

##### And thus, $A_{zero} = \{(n, j, 0) | j \in \{0, \ldots, n-1\}\}$ is non-empty:

```python
from proveit.logic import NotEquals
from proveit.logic.sets import EmptySet
NotEquals(A_zero, EmptySet).prove()
```

##### Next, we extend the non-emptiness conclusion to the union.

```python
from proveit import k
# notice here we MIGHT actually need the domain to be Interval(subtract(_n, zero), subtract(_n, one))
B_zero = SetOfAll(k, ExprTuple(k, _n, zero), domain=Interval(_n, subtract(_n, one)))
```

```python
from proveit.logic.sets import Union
NotEquals(Union(A_zero, B_zero), EmptySet).prove()
```

##### And then we extend the non-emptiness claim to the `UnionAll`

```python
from proveit import t
temp_exists_claim = Exists(t, NotEquals(
    Union(RoughBoundaryAVertices.definition().rhs.instance_expr.operands[0],
          RoughBoundaryAVertices.definition().rhs.instance_expr.operands[1]), EmptySet),
    domain=Interval(zero, _n))
```

```python
temp_exists_claim.conclude_via_example(zero)
```
