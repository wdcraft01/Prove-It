Given a surface code configuration $S$ consisting of data qubits, measurement qubits, and their interconnections,  and some error $E$ acting on that surface code, $\texttt{graph\_of}(E, S)$ denotes a generic _graphical_ representation of the error $E$ in the context of the graphical representation of $S$, made up now of edges and vertices. The generic `GraphOf` function will usually be replaced with a more specific graphical interpretation function such as `MergedBoundsGraphOfZError()`, etc. 

```python
GraphOf(E, S)
```

Given a surface code configuration $S$ consisting of data qubits, measurement qubits, and their interconnections, $\texttt{SiteSyndrome}(E, S)$ denotes the set of $X$ checks or site operators that signal one or more errors in the surface code in response (_i.e._, the set of $X$ site checks that “flip state” as a result of the given error $E$). Recall that such a site syndrome may be empty even in the presence of an error chain, such as in the case of a logical error, a logical $Z$, or any $X$-error chain.

```python
SiteSyndrome(E, S)
```

Given a surface code configuration $S$ consisting of data qubits, measurement qubits, and their interconnections, $\texttt{InteriorVertices}(GraphOf(E, S))$ denotes the set of interior vertices (_i.e._, non-boundary vertices) in the subgraph associated with component $E$, and $\texttt{BoundaryVertices}(GraphOf(E, S))$ denotes the set of boundary (non-interior) vertices, where “GraphOf” is some function producing a graphical representation of $E$ in the context of the graphical representation of $S$..

So, after choosing some convenient element of the set, how do we actually prove that element is in the set?

In our case, we choose the tuple $(n, 0, 0)$ and want to prove $(n,0,0) \in \{(n, j, t)\}_{j\in\{t,\ldots,n-1\}}$. This means that we have also chosen $t = 0$, so we'll carry along the assumption that $t = 0$.

To prove that $(n,0,0) \in \{(n, j, t)\}_{j\in\{t,\ldots,n-1\}}$, we can prove its definitional equivalent (the definition stemming from the fact that this membership is specifically a `SetOfAllMembership`):

```python
the_def = InSet(ExprTuple(_n, zero, zero), example_set_of_all).definition()
```

But apparently we'll need to establish a few more facts before this works, including some information about that constant $n$:

```python
from proveit import t
from proveit.logic import Equals
try:
    the_def.rhs.conclude_via_example(zero, assumptions=[Equals(t, zero)])
except Exception as the_exception:
    print(f"EXCEPTION: {the_exception}")
```

```python
from proveit import t
from proveit.logic import Equals
from proveit.numbers import zero
from proveit.physics.quantum.QEC.kitaev_planar_hadamard import (
        _n_in_integer, _n_ge_three)
InSet(zero, example_set_of_all.domain).prove(
    assumptions=[Equals(t, zero)])
```

Now we can derive the intermediate existence claim:

```python
exists_j_judgment = the_def.rhs.conclude_via_example(zero, assumptions=[Equals(t, zero)])
```

And then substitute in the definitional equivalent:

```python
the_def.sub_left_side_into(exists_j_judgment, assumptions=[Equals(t, zero)])
```

Then we can derive the “final” existence claim:

```python
_P_sub = Lambda(x, example_exists.instance_expr)
_x_sub = ExprTuple(_n, zero, zero) # we've now proven this is a valid example element
existence_by_example_single_inst = existence_by_example_single.instantiate(
            {P:_P_sub, x:_x_sub}, assumptions=[Equals(t, zero)])
```

Eventually then we'll want to be able to conclude that $A \cup B \ne \emptyset$ if we know that $A$ or $B$ is non-empty, and that $\bigcup_{i}A_{i} \ne \emptyset$ if we know there exists some $i$ such that $A_{i} \ne \emptyset$.

To do that more automatically:

(1) Thm(s) along the lines of: $\forall_{A,B | A \ne \emptyset}\big[A \cup B \ne \emptyset\big]$ and $\forall_{A|A \ne \emptyset}\bigcup_{i \in S}\big[A(i)\big]$

(2) Link this into the NotEquals() system, where NotEquals.side_effects() which checks if rhs (_i.e._, $\emptyset$) has a not_equals_side_effects attribute. And our EmptySet does indeed have such a not_equals_side_effects attribute, in which we can then conclude that the lhs has the Exists(x, x in lhs) property. This should already work as-is, and is not going to achieve the automaticity I was thinking about. 

(3) How to automate the conclusion that $A \cup B \ne \emptyset$ if we know that A (or B) is not $\emptyset$? In NotEquals.conclude(), the system checks if the rhs has a deduce_not_equal attribute. Perhaps we can add such a deduce_not_equal() method to EmptySet? Which could then check for special cases, like a union involving a non-empty set.
